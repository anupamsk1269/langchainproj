```python
import sys
import os
import importlib

sys.path.append(os.path.abspath(os.path.join(os.getcwd(), os.pardir)))
import config
importlib.reload(config)

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough
from langchain_openai import ChatOpenAI
from langchain_anthropic import ChatAnthropic

# =====================================================================
# 1. INITIALIZE COMPONENTS & THE TWO MODELS
# =====================================================================
openai_model = ChatOpenAI(model=config.DEFAULT_MODEL, api_key=config.OPENAI_API_KEY)
claude_model = ChatAnthropic(model="claude-sonnet-4-6", api_key=config.ANTHROPIC_API_KEY)
parser = StrOutputParser()

# --- Chain 1: The Generation Phase (OpenAI) ---
generation_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are an efficient {language} programmer. Write a clean function."),
    ("human", "Write a function to perform: {task}")
])
generation_chain = generation_prompt | openai_model | parser

# --- Chain 2: The Audit Phase (Claude) ---
# Note how this prompt expects BOTH the original task AND the generated_code!
audit_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are an expert security engineer. Review the provided code against the original task goal. Call out any performance bugs, edge case vulnerabilities, or missing safeguards."),
    ("human", "Original Goal: {task}\n\nWritten Code:\n{generated_code}")
])
audit_chain = audit_prompt | claude_model | parser

# =====================================================================
# 2. CONSTRUCT THE COMPLEX PIPELINE USING RUNNABLEPASSTHROUGH.ASSIGN()
# =====================================================================
# Let's read the data flow left-to-right:
# 1. Input dict enters.
# 2. .assign() kicks off generation_chain, capturing the result as "generated_code".
# 3. .assign() passes the MERGED dict {"language":..., "task":..., "generated_code":...} to audit_chain.
full_pipeline = (
    RunnablePassthrough.assign(generated_code=generation_chain) 
    | audit_chain
)

# =====================================================================
# 3. EXECUTE THE PIPELINE
# =====================================================================
inputs = {
    "language": "Python",
    "task": "Accept a file path, load user account metadata from a JSON profile, and update a database balance entry securely."
}

print("Executing complex multi-model pipeline...\n")
final_audit_report = full_pipeline.invoke(inputs)

print("=== Final Claude Security & Performance Audit ===")
print(final_audit_report)
```

```python
import warnings
warnings.filterwarnings("ignore", category=RuntimeWarning)

import sys
import os
import importlib

sys.path.append(os.path.abspath(os.path.join(os.getcwd(), os.pardir)))
import config
importlib.reload(config)

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_ollama import ChatOllama
from langchain_openai import ChatOpenAI

# Variables to log state tracking
active_route = "None"

# Define lifecycle callbacks for the primary chain
def on_primary_success(run_obj):
    global active_route
    active_route = "Primary Local Chain (Ollama)"

def on_primary_error(run_obj):
    print("   [Log] Primary Local Chain failed. Diverting traffic...")

# Define lifecycle callbacks for the backup chain
def on_backup_success(run_obj):
    global active_route
    active_route = "Backup Cloud Chain (OpenAI)"


# =====================================================================
# 1. CONSTRUCT PRIMARY WITH LISTENERS
# =====================================================================
# Simulating a broken local model tag again to force the failover
local_model = ChatOllama(model="mistral-broken-tag-simulation", base_url="http://localhost:11434")
local_prompt = ChatPromptTemplate.from_messages([("human", "Explain telemetry in one sentence.")])

# Attach listeners to the primary pipeline block
primary_chain = (
    (local_prompt | local_model | StrOutputParser())
    .with_listeners(on_end=on_primary_success, on_error=on_primary_error)
)

# =====================================================================
# 2. CONSTRUCT BACKUP WITH LISTENERS
# =====================================================================
cloud_model = ChatOpenAI(model=config.DEFAULT_MODEL, api_key=config.OPENAI_API_KEY, temperature=0.3)
cloud_prompt = ChatPromptTemplate.from_messages([("human", "Explain telemetry in one sentence.")])

# Attach listeners to the backup pipeline block
backup_chain = (
    (cloud_prompt | cloud_model | StrOutputParser())
    .with_listeners(on_end=on_backup_success)
)


# =====================================================================
# 3. UNIFY AND EXECUTE
# =====================================================================
resilient_system_chain = primary_chain.with_fallbacks([backup_chain])

print("Invoking resilient system...")
output = resilient_system_chain.invoke({})

print("\n=== Audit Metrics ===")
print(f"Successfully Executed Route: {active_route}")
print(f"Response Content: {output}")
```

```python
import warnings
warnings.filterwarnings("ignore", category=DeprecationWarning)
warnings.filterwarnings("ignore", category=RuntimeWarning)

import sys
import os
import importlib

sys.path.append(os.path.abspath(os.path.join(os.getcwd(), os.pardir)))
import config
importlib.reload(config)

from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableLambda

# =====================================================================
# 1. Setup the AUDITED Image Generator Function
# =====================================================================
def create_robust_generator(api_key: str):
    """Factory function to build a function with explicit error/None handling"""
    from openai import OpenAI
    client = OpenAI(api_key=api_key)
    
    def generate_image(prompt_text: str) -> str:
        """Invokes modern image engine with internal fallback logic"""
        try:
            print("   [OAI API] Submitting audited generation request...")
            # Attempt to invoke the standardized generation successor model
            response = client.images.generate(
                model="gpt-image-2",
                prompt=prompt_text,
                n=1,
                size="1024x1024"
            )
            
            # CRITICAL AUDIT: Check if the API returned a valid URL or silent None
            generated_url = response.data[0].url
            
            if generated_url is None:
                # API physically returned None (not an error, but an invalid state for us)
                print("   [OAI API] Generation call failed silently (None returned). Triggering Failsafe...")
                raise ValueError("API returned None.")
            
            return generated_url
            
        except Exception as e:
            # =====================================================================
            # 2. DEFINING THE STABLE FALLBACK BEHAVIOR
            # =====================================================================
            # In a production app, this would be a local 'image_not_available.png' asset.
            # For this notebook, we'll use a reliable public LangChain asset URL.
            print(f"   [OAI API ERROR]: {e}")
            print("   [Audit] Generation call failed or returned None. Failsafe activated...")
            
            # Using a reliable, stable placeholder URL so the chain *always* returns a string.
            fallback_url = "https://raw.githubusercontent.com/langchain-ai/langchain/master/docs/static/img/langchain_logo.png"
            return fallback_url

    return generate_image

# 3. Construct the full LCEL Pipeline
# We instantiate the audited function factory.
robust_generator = create_robust_generator(config.OPENAI_API_KEY)
# We wrap the audited function inside a LangChain Runnable function call.
image_generator = RunnableLambda(lambda x: robust_generator(x))

expansion_prompt = PromptTemplate.from_template(
    "Expand this idea into a detailed, creative image prompt: {image_desc}. "
    "Focus on style, lighting, composition, and mood."
)
llm = ChatOpenAI(model="gpt-4o-mini")

generation_chain = (
    expansion_prompt | llm | StrOutputParser() | image_generator
)

# 4. Invoke the chain to synthesize the image
image_desc = "A cozy, vintage library nook with a soft armchair and warm ambient lighting"
print(f"Synthesizing image for: '{image_desc}'...")

# Calling .invoke() now returns either a fresh URL or a stable placeholder string.
final_url = generation_chain.invoke({"image_desc": image_desc})
print(f"Resulting Image Output (Robust URL): {final_url}")
```

```python
import warnings
warnings.filterwarnings("ignore", category=DeprecationWarning)
warnings.filterwarnings("ignore", category=RuntimeWarning)
# Pytorch (Stable Diffusion engine) has many internal warnings on CPUs
warnings.filterwarnings("ignore", message="Using CPU for execution*")

import sys
import os
import importlib
import torch

sys.path.append(os.path.abspath(os.path.join(os.getcwd(), os.pardir)))
import config
importlib.reload(config)

from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI
# New standard for local Hugging Face interactions
from langchain_huggingface import HuggingFacePipeline
# Diffusers handles the raw Stable Diffusion math
from diffusers import StableDiffusionPipeline


# =====================================================================
# 1. SETUP THE LOCAL STABLE DIFFUSION GENERATOR
# =====================================================================
# This loads optimized Stable Diffusion weights (about 4GB) to your computer's CPU
model_id = "runwayml/stable-diffusion-v1-5"
print(f"Loading '{model_id}' weights locally onto CPU...")
print("-> (This may take a few minutes on the first run, and requires memory)...")

# Construct the diffusers pipeline natively
# For local hardware constraints, we force CPU execution. If you have a powerful GPU (like Mac M1/M2), 
# you could swap this for device="mps" or device="cuda".
hf_pipe = StableDiffusionPipeline.from_pretrained(
    model_id, 
    torch_dtype=torch.float32  # Using standard precision for CPU compatibility
)
hf_pipe = hf_pipe.to("cpu")

# Wrap the diffusers object as a custom LangChain Runnable
def local_image_synthesis(expanded_prompt: str) -> str:
    """Invokes local Stable Diffusion with the LLM's expanded output"""
    print(f"   [Local HF] Synthesizing image from expanded prompt (CPU generation)...")
    print("   [Note] This process is heavy; please allow 30-120 seconds per generation.")
    
    # Execute image generation
    output_object = hf_pipe(
        prompt=expanded_prompt,
        height=512,  # Using smaller dimensions for CPU speed (512x512 is native)
        width=512,
        num_inference_steps=20,  # Limits steps to save generation time
        guidance_scale=7.5
    )
    
    # Save the resulting PIL image to disk
    image_asset = output_object.images[0]
    local_filename = "generated_stable_diffusion.png"
    image_asset.save(local_filename)
    return local_filename # Return the path to the fresh image, not just text/None!

from langchain_core.runnables import RunnableLambda
image_generator = RunnableLambda(local_image_synthesis)


# =====================================================================
# 2. BUILD AND EXECUTE THE ENTIRE LCEL CHAIN
# =====================================================================
expansion_prompt = PromptTemplate.from_template(
    "You are a cinematic prompt engineer. Expand this simple concept into a short, visually "
    "descriptive prompt for an image generator (like Stable Diffusion). Focus on art style, "
    "lighting, and textures. Under 50 words.\nConcept: {image_desc}"
)

expansion_llm = ChatOpenAI(model=config.DEFAULT_MODEL, api_key=config.OPENAI_API_KEY)

# Data Flow: Dict -> Prompt -> LLM -> Text String -> Local HF Generator -> Image Filename
text_to_image_chain = expansion_prompt | expansion_llm | StrOutputParser() | image_generator

user_concept = "A cozy, vintage library nook with warm ambient lighting"
print(f"\nExecuting local generation pipeline for: '{user_concept}'...")

try:
    final_output_path = text_to_image_chain.invoke({"image_desc": user_concept})
    print("\n=== Local Image Synthesis Successful ===")
    print(f"Image saved locally to disk as: {final_output_path}")
except Exception as e:
    print(f"\nLocal generation failed: {e}")
```
