```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI

# 1. Load environment variables from the .env file
load_dotenv(dotenv_path="../.env")

# 2. Initialize the OpenAI LLM client
# (It automatically looks for the OPENAI_API_KEY environment variable)
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)

# 3. Test a quick invocation
try:
    response = llm.invoke("Hello! Confirming our connection is live.")
    print("Success! Response from LLM:")
    print(response.content)
except Exception as e:
    print(f"Error connecting: {e}")
```

```python
# 1. Install tiktoken directly from your notebook cell
!pip install tiktoken

import tiktoken

# 2. Get the specific tokenizer used by modern GPT models (o1, gpt-4o, gpt-4o-mini)
encoding = tiktoken.get_encoding("cl100k_base")

# 3. Test text
text = "LangChain makes AI engineering incredibly powerful."

# 4. Convert text to token IDs
tokens = encoding.encode(text)
print(f"Token IDs: {tokens}")
print(f"Total number of tokens: {len(tokens)}")

# 5. Let's see exactly what text chunks those IDs represent
text_chunks = [encoding.decode([t]) for t in tokens]
print(f"Token Chunks: {text_chunks}")
```

```python
from langchain_openai import OpenAIEmbeddings

# 1. Initialize the OpenAI Embeddings model
embeddings_model = OpenAIEmbeddings(model="text-embedding-3-small")

# 2. Convert a phrase into an embedding vector
text = "AI engineering"
vector = embeddings_model.embed_query(text)

# 3. Inspect the vector
print(f"Vector type: {type(vector)}")
print(f"Number of dimensions (length of the list): {len(vector)}")
print(f"First 5 numbers in the vector: {vector[:5]}")
```
